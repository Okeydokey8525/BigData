```python
from pathlib import Path
import pandas as pd
import numpy as np
```

Đọc dữ liệu


```python
DATA_PATH=Path("../data/raw/flight_data_2024.csv")
df=pd.read_csv(DATA_PATH)

print("Đọc Dữ Liệu Thành Công")
print(f"Số dòng:{df.shape[0]}")
print(f"Số cột:{df.shape[1]}")
```

Xem dữ liệu ban đầu


```python
pd.set_option("display.max_columns", None)
display(df[(df["cancelled"] == 1) & (df["late_aircraft_delay"] > 0)].head(5))
```

Danh sach thuộc tính


```python
column_info = pd.DataFrame({
    "STT": range(1, len(df.columns) + 1),
    "Thuộc tính": df.columns,
    "Kiểu dữ liệu": df.dtypes.astype(str).values,
    "Số giá trị khác nhau": [
        df[column].nunique(dropna=True)
        for column in df.columns
    ]
})

display(column_info)
```

Kiểm tra missing


```python
missing_info = pd.DataFrame({
    "Thuộc tính": df.columns,
    "Số missing": df.isna().sum().values,
    "Tỷ lệ missing (%)": (
        df.isna().mean().values * 100
    ).round(2)
})

missing_info = missing_info[
    missing_info["Số missing"] > 0
].sort_values(
    "Số missing",
    ascending=False
)

display(missing_info)
```

Kiểm tra duplicate


```python
duplicate_count=df.duplicated().sum()
print(f"Số dòng bị trùng:{duplicate_count:,}")
```

Thống kê biến số


```python
numeric_cols=df.select_dtypes(include=np.number).columns
numeric_summary = df[numeric_cols].describe().T
display(numeric_summary)
```

Thống kê biến phân loại


```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

categorical_summary = pd.DataFrame({
    "Thuộc tính": categorical_columns,
    "Số giá trị khác nhau": [
        df[column].nunique(dropna=True)
        for column in categorical_columns
    ],
    "Missing": [
        df[column].isna().sum()
        for column in categorical_columns
    ]
})

display(categorical_summary)
```

Kiểm tra 5 nguyên nhân delay


```python
CAUSE_COLUMNS = [
    "carrier_delay",
    "weather_delay",
    "nas_delay",
    "security_delay",
    "late_aircraft_delay"
]

cause_summary = pd.DataFrame({
    "Số missing": df[CAUSE_COLUMNS].isna().sum(),
    "Số bằng 0": (df[CAUSE_COLUMNS] == 0).sum(),
    "Số > 0": (df[CAUSE_COLUMNS] > 0).sum(),
    "Giá trị lớn nhất": df[CAUSE_COLUMNS].max(),
    "Giá trị trung vị": df[CAUSE_COLUMNS].median()
})

display(cause_summary)
```

Kiểm tra một chuyến có nhiều nguyên nhân hay không


```python
cause_presence = df[CAUSE_COLUMNS].gt(0)

number_of_causes = cause_presence.sum(axis=1)

cause_count_summary = (
    number_of_causes
    .value_counts()
    .sort_index()
    .rename_axis("Số nguyên nhân")
    .reset_index(name="Số chuyến bay")
)

cause_count_summary["Tỷ lệ (%)"] = (
    cause_count_summary["Số chuyến bay"]
    / len(df)
    * 100
).round(2)

display(cause_count_summary)
```

Số Chuyến Bị Huỷ

```python
cancelled_count=(df["cancelled"]==1).sum()
total_flights=len(df)

cancelled_percentage=cancelled_count/total_flights*100
print(f"Số chuyến bay bị hủy: {cancelled_count:,}")
print(f"Tổng số chuyến bay: {total_flights:,}")
print(f"Tỷ lệ chuyến bay bị hủy: {cancelled_percentage:.2f}%")
```

xác định số dòng thực sự bị ảnh hưởng

```python
missing_columns = [
    "dep_time",
    "dep_delay",
    "taxi_out",
    "wheels_off",
    "taxi_in",
    "wheels_on",
    "arr_time",
    "arr_delay",
    "actual_elapsed_time",
    "air_time"
]

affected_rows = df[missing_columns].isna().any(axis=1).sum()

print(f"Số chuyến bay có ít nhất một missing: {affected_rows:,}")
print(f"Tỷ lệ: {affected_rows / len(df) * 100:.2f}%")
```

Lưu kết quả khảo sát


```python
REPORT_PATH = Path("../reports/data_understanding_summary.txt")

with open(REPORT_PATH, "w", encoding="utf-8") as f:
    f.write(f"Số dòng: {df.shape[0]:,}\n")
    f.write(f"Số cột: {df.shape[1]}\n")
    f.write(f"Số dòng trùng hoàn toàn: {duplicate_count:,}\n\n")

    f.write("=== MISSING VALUES ===\n")
    f.write(missing_info.to_string(index=False))
    f.write("\n\n")

    f.write("=== CAUSE SUMMARY ===\n")
    f.write(cause_summary.to_string())
    f.write("\n\n")

    f.write("=== NUMBER OF CAUSES PER FLIGHT ===\n")
    f.write(cause_count_summary.to_string(index=False))

print(f"Đã lưu báo cáo: {REPORT_PATH}")
```
