```python
import pandas as pd
from pathlib import Path


# ============================================================
# 1. Đọc dữ liệu sau bước Data Cleaning
# ============================================================

INPUT_PATH = "../data/interim/flight_data_7m_cleaned.csv"

df = pd.read_csv(INPUT_PATH)

print("Kích thước dữ liệu ban đầu:", df.shape)


# ============================================================
# 2. Features được giữ lại
# ============================================================

FEATURE_COLUMNS = [
    "month",
    "day_of_month",
    "day_of_week",
    "fl_date",
    "op_unique_carrier",
    "op_carrier_fl_num",
    "origin",
    "dest",
    "crs_dep_time",
    "crs_arr_time",
    "crs_elapsed_time",
    "distance"
]


# ============================================================
# 3. 5 Target cần dự đoán
# ============================================================

TARGET_COLUMNS = [
    "carrier_delay",
    "weather_delay",
    "nas_delay",
    "security_delay",
    "late_aircraft_delay"
]


# ============================================================
# 4. Các biến bị loại
# ============================================================

REMOVED_COLUMNS = [
    "year",
    "dep_time",
    "dep_delay",
    "taxi_out",
    "wheels_off",
    "wheels_on",
    "taxi_in",
    "arr_time",
    "arr_delay",
    "cancelled",
    "cancellation_code",
    "diverted",
    "actual_elapsed_time",
    "air_time",
    "origin_city_name",
    "origin_state_nm",
    "dest_city_name",
    "dest_state_nm"
]


# ============================================================
# 5. Kiểm tra các cột cần thiết
# ============================================================

REQUIRED_COLUMNS = FEATURE_COLUMNS + TARGET_COLUMNS

missing_columns = [
    column
    for column in REQUIRED_COLUMNS
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Không tìm thấy các cột: {missing_columns}"
    )


# ============================================================
# 6. Tạo dataset sau Feature Selection
# ============================================================

df_selected = df[REQUIRED_COLUMNS].copy()


# ============================================================
# 7. Kiểm tra kết quả
# ============================================================

print("\n" + "=" * 60)
print("KẾT QUẢ FEATURE SELECTION")
print("=" * 60)

print(f"Số feature : {len(FEATURE_COLUMNS)}")
print(f"Số target  : {len(TARGET_COLUMNS)}")
print(f"Tổng biến  : {len(REQUIRED_COLUMNS)}")
print(f"Số dòng    : {len(df_selected)}")

print("\n--- FEATURES ĐƯỢC GIỮ ---")

for i, column in enumerate(FEATURE_COLUMNS, start=1):
    print(f"{i:2}. {column}")

print("\n--- TARGETS ---")

for i, column in enumerate(TARGET_COLUMNS, start=1):
    print(f"{i:2}. {column}")

print("\n--- CÁC BIẾN BỊ LOẠI ---")

for i, column in enumerate(REMOVED_COLUMNS, start=1):
    print(f"{i:2}. {column}")


# ============================================================
# 8. Kiểm tra missing values
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUES SAU FEATURE SELECTION")
print("=" * 60)

missing_summary = (
    df_selected.isnull()
    .sum()
    .to_frame(name="missing_count")
)

missing_summary["missing_percentage"] = (
    missing_summary["missing_count"]
    / len(df_selected)
    * 100
)

missing_summary = missing_summary[
    missing_summary["missing_count"] > 0
]

if missing_summary.empty:
    print("Không còn missing values.")
else:
    print(missing_summary)


# ============================================================
# 9. Kiểm tra danh sách cột cuối cùng
# ============================================================

print("\n" + "=" * 60)
print("DANH SÁCH CỘT CUỐI CÙNG")
print("=" * 60)

print(df_selected.columns.tolist())


# ============================================================
# 10. Tạo thư mục output
# ============================================================

OUTPUT_PATH = Path("../data/processed/flight_data_7m_selected.csv")

OUTPUT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 11. Lưu dataset sau Feature Selection
# ============================================================

df_selected.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 60)
print("LƯU FILE")
print("=" * 60)

print(f"Đã lưu tại: {OUTPUT_PATH}")
```
