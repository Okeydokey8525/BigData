Đọc dữ liệu


```python
from pathlib import Path
import pandas as pd

# DATA_PATH = Path("../data/raw/flight_data_2024_sample.csv")
# OUTPUT_PATH = Path("../data/interim/flight_data_cleaned.csv")
DATA_PATH = Path("../data/raw/flight_data_2024.csv")
OUTPUT_PATH = Path("../data/interim/flight_data_7m_cleaned.csv")



df = pd.read_csv(DATA_PATH)

print(f"Kích thước dữ liệu đầu vào: {df.shape}")
```

2. Loại dòng trùng


```python
before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

print(f"Số dòng đã loại: {before - len(df):,}")
print(f"Kích thước sau khi loại dòng trùng: {df.shape}")
```

3. Loại chuyến bay bị hủy


```python
before = len(df)

df = df[df["cancelled"] == 0].copy()

removed_cancelled = before - len(df)

print(f"Số chuyến bay bị hủy đã loại: {removed_cancelled:,}")
print(f"Kích thước sau khi loại chuyến bay bị hủy: {df.shape}")
```

4. Xử lý missing


```python
MISSING_COLUMNS = [
    "dep_time",
    "dep_delay",
    "taxi_out",
    "wheels_off",
    "wheels_on",
    "arr_time",
    "arr_delay",
    "actual_elapsed_time",
    "air_time",
    "op_carrier_fl_num",
    "crs_elapsed_time"
]
before = len(df)

df = df.dropna(subset=MISSING_COLUMNS).reset_index(drop=True)

print(f"Số dòng bị loại do missing: {before - len(df):,}")
print(f"Kích thước sau xử lý missing: {df.shape}")
```

5. Xử lý fl_date


```python
df["fl_date"] = pd.to_datetime(
    df["fl_date"],
    errors="coerce"
)
```

6. Xử lý các thuộc tính giờ


```python
TIME_COLUMNS = [
    "crs_dep_time",
    "dep_time",
    "wheels_off",
    "wheels_on",
    "crs_arr_time",
    "arr_time"
]

for col in TIME_COLUMNS:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    hours = df[col] // 100
    minutes = df[col] % 100

    df[col] = hours * 60 + minutes
```

Lưu dữ liệu sau Cleaning


```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã lưu dữ liệu sau Cleaning: {OUTPUT_PATH}")
print(f"Kích thước cuối Bước 2: {df.shape}")
```
